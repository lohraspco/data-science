Basic LLM Chain

```python
# Use this to show prompt templating and chaining with an LLM.
from langchain.llms import OpenAI
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate

prompt = PromptTemplate.from_template("What is a good name for a company that makes {product}?")
llm = OpenAI(temperature=0.7)
chain = LLMChain(llm=llm, prompt=prompt)

result = chain.run(product="saffron beer")
print(result)
```

RAG

```python
from langchain.vectorstores import FAISS
from langchain.embeddings import OllamaEmbeddings
from langchain.chains import RetrievalQA
from langchain.llms import Ollama

db = FAISS.load_local("my_vector_db", OllamaEmbeddings(model="nomic-embed-text"))
retriever = db.as_retriever()

qa_chain = RetrievalQA.from_chain_type(
    llm=Ollama(model="gemma:4b"),
    retriever=retriever,
    chain_type="stuff"
)
print(qa_chain.run("What did the user say about pricing?"))
```

Conversational Memory Chain

useful for agents or assistants that need to retain context.

```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory
from langchain.llms import Ollama

memory = ConversationBufferMemory()
chat_chain = ConversationChain(llm=Ollama(model="llama3:8b"), memory=memory)

chat_chain.run("Hi, I'm Matt.")
chat_chain.run("What's my name?")
```

Tool-Using Agent

Good for showing reasoning, planning, and tool use via agents.

```python
from langchain.agents import initialize_agent, load_tools
from langchain.agents.agent_types import AgentType
from langchain.llms import OpenAI

tools = load_tools(["serpapi", "llm-math"], llm=OpenAI())
agent = initialize_agent(tools, OpenAI(), agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION)

response = agent.run("What's the current price of gold and what's 10% of it?")
print(response)
```

Custom Document Loader + Embedding

Great for explaining ingestion pipeline before RAG — loading, chunking, embedding, vector DB.

```python
from langchain.document_loaders import TextLoader
from langchain.text_splitter import CharacterTextSplitter
from langchain.vectorstores import Chroma
from langchain.embeddings import HuggingFaceEmbeddings

loader = TextLoader("privacy_policy.txt")
docs = loader.load()

splitter = CharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = splitter.split_documents(docs)

db = Chroma.from_documents(chunks, HuggingFaceEmbeddings())
```
