```python
import xgboost as xgb
from sklearn.datasets import make_classification
from collections import Counter

# Create an imbalanced dataset
X, y = make_classification(n_samples=1000, n_features=10, n_informative=5,
                           n_redundant=0, n_repeated=0, n_classes=2,
                           n_clusters_per_class=1, weights=[0.9, 0.1],
                           flip_y=0, random_state=42)

# Calculate scale_pos_weight
counter = Counter(y)
scale_pos_weight_value = counter[0] / counter[1] # Majority class count / Minority class count

# Initialize and train XGBoost classifier
xgb_model = xgb.XGBClassifier(objective='binary:logistic', eval_metric='auc',
                              scale_pos_weight=scale_pos_weight_value,
                              random_state=42)
xgb_model.fit(X, y)
```

```python
from langchain.schema import Document
docs = [
    Document(page_content="Ollama is a tool that runs LLMs locally on your machine."),
    Document(page_content="Gemma is an open LLM developed by Google. It comes in 3233B and 34334343343B versions."),
    Document(page_content="FAISS is used to perform similarity search over dense vectors."),
]
```

```python
# 🔹 Split docs if needed (optional here)
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(chunk_size=300, chunk_overlap=20)
splitted_docs = text_splitter.split_documents(docs)
```

```python
from langchain.embeddings import OllamaEmbeddings
from langchain.vectorstores import FAISS
embedding = OllamaEmbeddings(model="nomic-embed-text")

# 🔹 Store vectors in FAISS
vectorstore  = FAISS.from_documents(splitted_docs, embedding)
```

```python
from langchain.chat_models import ChatOllama
llm = ChatOllama(model="gemma3:4b")
```

```python
# 🔹 Create a RetrievalQA chain
qa = RetrievalQA.from_chain_type(llm=llm, retriever=vectorstore.as_retriever())
```

```python
# 🔹 Ask a question!
query = "What is Gemma?"
response = qa.run(query)

print(f"Q: {query}")
print(f"A: {response}")
```
