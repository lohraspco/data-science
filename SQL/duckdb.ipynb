```python
import duckdb
```

```python
import pandas as pd
con1 = duckdb.connect()
df = pd.DataFrame({"a": [1,2,3], "b": [4,5,6]})
con1.register("df", df)

con1.sql("SELECT a, b, a+b AS c FROM df").df()
```
