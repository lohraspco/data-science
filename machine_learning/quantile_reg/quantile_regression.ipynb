- (OLS) regression focuses on the mean of the distribution
- Quantile regression is a statistical technique used to estimate the conditional quantiles of a response variable (y) given predictors (x).
    - Quantile regression is great when the relationship between variables changes across the outcome distribution, unlike OLS which assumes a constant effect.
    - handling non-normal data distributions
    - heteroskedasticity (varying variance)
    - when outliers might skew the mean. 
    - machine learning (robust regression).
    - No Distribution Assumptions
    - For a quantile τ, minimizes the sum of asymmetrically weighted absolute residuals
    - Cons: Computationally more intensive; interpretation is quantile-specific.

```python
import numpy as np
import statsmodels.formula.api as smf
import pandas as pd

np.random.seed(42)  # For reproducibility
n = 100
x = np.random.uniform(0, 10, n)
error = np.random.normal(0, x, n)
y = 2 + 3 * x + error
df = pd.DataFrame({'x': x, 'y': y})

quantiles = [0.25, 0.50, 0.75]
models = {}
predictions = {}
for q in quantiles:
    # Fit model: y ~ x
    model = smf.quantreg('y ~ x', df).fit(q=q)
    models[q] = model
    
    # Predict on original x values
    df[f'pred_q{q}'] = model.predict(df['x'])
    predictions[q] = model.params
print("Quantile Regression Results:")
print("-" * 50)
for q in quantiles:
    params = models[q].params
    print(f"τ = {q:.2f}: Intercept = {params['Intercept']:.3f}, Slope = {params['x']:.3f}")
```
