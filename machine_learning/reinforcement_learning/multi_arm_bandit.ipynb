```python
import torch 
import random

# Simulate 10 arms, each with a fixed but unknown reward probability
k = 10

# 1- Setuo environment and bandits
true_mean = torch.randn(k) # unknown to the agent
def pull_arm(arm):
    """Return a reward (stochastic) for pulling a given arm"""
    return torch.rand(1).item() + true_mean[arm].item()

# 2- Setup Epsilon strategy
class EpsilonGreedyAgent:
    """strategy"""
    def __init__(self, k, epsilon=0.1):
        self.k = k
        self.epsilon = epsilon
        self.counts = torch.zeros(k)     # How many times each arm was pulled
        self.q_values = torch.zeros(k)   # Estimated value of each arm

    def select_arm(self): 
        if random.random() < self.epsilon:
            # print("explor")
            return random.randint(0, self.k - 1)  # Explore
        else:
            # print("exploit")
            return torch.argmax(self.q_values).item()  # Exploit

    def update(self, arm, reward):
        self.counts[arm] += 1
        alpha = 1 / self.counts[arm]  # incremental update
        self.q_values[arm] += alpha * (reward - self.q_values[arm])


```

```python
# Step 3: Training Loop

agent = EpsilonGreedyAgent(k=10, epsilon=0.3)
rewards = []
for step in range(1000):
    arm = agent.select_arm()
    reward = pull_arm(arm)
    agent.update(arm, reward)
    rewards.append(reward)
```

```python
print(f"Average reward: {sum(rewards) / len(rewards):.2f}")
print(f"Estimated values: {agent.q_values}")
print(f"True means:       {true_mean}")
```
