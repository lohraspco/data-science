Interaction Cycle: The typical interaction with a Gymnasium environment involves:
- Creation: env = gymnasium.make("EnvironmentName-vX")
- Reset: observation, info = env.reset() (to start a new episode)
- Step: observation (s_(t+1)), reward, (episode) terminated, truncated (timelimit or if an agent go out of bounds), info = env.step(action) (to take an action and observe the consequences)
- Closing: env.close() (to release resources)

#  Lunar Lander

problem: being able to land the Lunar Lander to the Landing Pad correctly by controlling left, right and main orientation engine

```python
env = gym.make("LunarLander-v3")
env.reset()
```

```python
print("_____OBSERVATION SPACE_____ \n")
print("Observation Space Shape", env.observation_space.shape)
columns = [
    'x',
    'y',
    'speed_x',
    'speed_y',
    'angle',
    'angular_speed',
    'left_leg_contact',
    'right_leg_contact'
]
pd.Series (env.observation_space.sample(), index= columns) 
```

```python
print("\n _____ACTION SPACE_____ \n")
print("Action Space Shape", env.action_space.n)
actions = ["Do nothing", "Fire left orientation engine", "Fire the main engine","Fire right orientation engine"]
a = env.action_space.sample()
print("Action Space Sample", a, actions[a]) 
```

```python
# create a vectorized environment of 16 environments
# (a method for stacking multiple independent environments into a single environment)
# this way, **we'll have more diverse experiences during the training.**
env = make_vec_env('LunarLander-v3', n_envs=16)
```

```python
model = PPO("MlpPolicy", env, verbose=1)
model.learn(total_timesteps=100000)  # you can increase the number of timesteps to improve performance
```

```python
import gymnasium as gym

# Create an environment
env = gym.make("CartPole-v1")

# Reset the environment to start a new episode
observation, info = env.reset(seed=42)

# Interact with the environment for a few steps
for _ in range(10):
    action = env.action_space.sample()  # Take a random action
    observation, reward, terminated, truncated, info = env.step(action)

    if terminated or truncated:
        observation, info = env.reset()

# Close the environment
env.close()
```
