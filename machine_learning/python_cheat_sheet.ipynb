```python
import itertools
import operator

# Accumulation with an initial value
data_with_initial = [1, 2, 3]
accumulated_with_initial = list(itertools.accumulate(data_with_initial,operator.mul, initial=10))
print(f"Accumulated with initial: {accumulated_with_initial}") # Output: [10, 11, 13, 16]
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Calculate the sum of elements using reduce
sum_of_numbers = reduce(lambda x, y: x + y, numbers)
print(f"Sum: {sum_of_numbers}")  # Output: Sum: 15

# Find the maximum element using reduce
max_number = reduce(lambda x, y: x if x > y else y, numbers)
print(f"Maximum: {max_number}")  # Output: Maximum: 5
```

# Decorator

 Decorators in Python are a type of higher-order function because they take a function as input, modify it, and return a new function that extends or changes its behavio


```python
class bcolors:
    HEADER = '\033[95m'
    OKBLUE = '\033[94m'
    OKCYAN = '\033[96m'
    OKGREEN = '\033[92m'
    WARNING = '\033[93m'
    FAIL = '\033[91m'
    ENDC = '\033[0m'
    BOLD = '\033[1m'
    UNDERLINE = '\033[4m'

def print_colors(func):
    def wrapper(*args, **kwargs):
        print("before calling the function")
        a = func(*args, **kwargs)
        print(f"{bcolors.FAIL} {a} {bcolors.ENDC}")
        return a
    return wrapper

@print_colors
def say_hello(name):
    print(f"Hello, {name}!")
    return "Hello"
say_hello("Alice")
```

Generators 


```python
def count_up_to(max_num):
    n = 0
    while n < max_num:
        yield n
        n += 1

# Using the generator
my_generator = count_up_to(5)

print(next(my_generator))  # Output: 0
print(next(my_generator))  # Output: 1

for num in my_generator:
    print(num)  # Output: 2, 3, 4
```
