# Causation

- Correlation is a statistical relationship
- Causation means a change in one variable directly produces a change in the other.

Three pillars of causal inference

- **randomization** → removes bias by design 
- **confounders** → introduces bias if unaddressed ( that might influence both the treatment and the outcome  making it hard to tell if the treatment really caused the outcome. A confounder is a hidden variable that explains away the observed relationship. Ignoring it can make correlation look like causation.) Example: You observe that users who click ads tend to buy more products. You think: "Clicking ads causes purchases." But a confounder could be the Purchase intent — people already intending to buy are more likely to click ads and purchase.
So the real driver is not the ad click, but the user's intent — the confounder.

- **temporal ordering** (the cause must clearly happen before the effect)

How to Test for Causality:
- A/B Testing (Randomized Controlled Trial): Gold standard
- Natural Experiments (Instrumental variables, regression discontinuity)
- Time-series intervention models
- Granger Causality for temporal data
- Structural Causal Models (SCM) or Do-calculus (advanced)


Example:You find that users who watch more videos on Facebook tend to spend more money on ads.Does watching videos cause users to spend more?
Maybe. But maybe:
- Both are caused by a third factor (e.g., high engagement level)
- It’s reverse causation (spending more leads to targeted content)
- It’s just coincidence or seasonality

Bad inference:“People who use our new feed format click more. Let’s roll it out to all users.”
- Without an A/B test or natural experiment, you can’t be sure the feed format causes the clicks — maybe only highly active users adopted the new feed.

| Step             | Key Points                                                                 |
|------------------|-----------------------------------------------------------------------------|
| Design           | Define hypothesis, metric, required power/sample size                      |
| Randomization    | Assign users randomly to control/treatment                                 |
| Metric Selection | Choose primary & guardrail metrics                                         |
| Analysis         | T-test, Z-test, non-parametric test if needed                              |
| Pitfalls         | Peeking early; not accounting for multiple comparisons; novelty bias       |



 # Common Pitfalls to Know
- P-value ≠ Probability that H₀ is true
- Correlation ≠ Causation
- Survivorship bias: Ignoring missing data that dropped out
- Simpson’s Paradox: Trend reverses when data is aggregated
- Overfitting: High training accuracy, low generalization



# Propensity Score Matching (PSM)

Propensity Score Matching is a statistical technique used to reduce selection bias in observational studies by matching treated and untreated units (e.g., users) with similar propensity scores — the probability of receiving the treatment given observed characteristics.

Step 1: Estimate the Propensity Score
We use logistic regression to model the probability of being in the treatment group as a function of observed characteristics. The model is typically specified as: 
P(Treatment = 1 | X) = logit⁻¹(β₀ + β ₁X₁ + β₂X₂ + ... + βₖXₖ) where X represents observed covariates.

Step 2: Match Treated & Untreated Units
Use nearest neighbor, caliper, or kernel matching to pair users with similar scores:
- Treated user A (score = 0.73)
- Matched control user B (score = 0.74)

Step 3: Compare Outcomes
Now compare the outcome of interest (e.g., CTR, retention) between matched pairs.


PSM is a powerful tool used when you can’t run an A/B test but still want to estimate causal effects from observational data.

Example: "Let’s find users who did receive a treatment and match them with users who didn’t, but who looked similar beforehand."
Example:
- Treated group: Users who adopted a new UI
- Untreated group: Users who didn’t
- Confounder: Prior engagement (users who adopted might already be more active)

If you match users from both groups based on their propensity to adopt, you reduce the bias from those confounders and can better estimate the causal effect of adoption.


# Metrics

Taxonomy of Metrics
- Simple: Easy to understand and calculate, and people should be able to remember and discuss it easily.
- Clear: The definition is explicit and there is no ambiguity in interpretation.
- Actionable: The metric can be moved by improving products, and not easily gamed (making you feel like you are getting results even though it offers no insights into actual business health or growth)


## Success Metrics

These metrics are designed to directly measure if the feature is achieving its intended purpose and providing value to the user.
1.	Adoption Rate:
- What it is: The percentage of active users who use the new feature at least once.
- Why it matters: This is the next logical step after the click. It confirms that users aren't just clicking out of curiosity but are actually attempting to engage with the feature. You can track this as a weekly or monthly rate to see if adoption is growing.
2.	Completion Rate:
- What it is: The percentage of users who start the feature's workflow and successfully complete it.
- Why it matters: This is the most direct measure of the feature's effectiveness. A high completion rate suggests the feature is intuitive and functional; a low rate suggests friction, bugs, or a confusing user flow.
3.	Repeat Usage Rate (or Stickiness):
- What it is: The percentage of users who use the feature multiple times over a given period (e.g., daily, weekly).
- Why it matters: This is a powerful signal of value. A high repeat usage rate indicates the feature has become a regular part of a user's workflow, solving a persistent problem for them.
4.	Time to Value (TTV):
- What it is: The average time it takes for a user to achieve their first successful outcome using the feature.
- Why it matters: If the feature is meant to simplify a task, this metric will show if it's succeeding. A shorter TTV compared to the previous method indicates improved efficiency and a better experience.
5.	Qualitative Feedback (e.g., CSAT, NPS):
- What it is: A measure of user sentiment through in-app surveys, post-usage pop-ups, or interviews. You can ask questions like, "How much did this feature help you?" or track the overall Net Promoter Score (NPS) for users who've tried the feature versus those who haven't.
- Why it matters: This provides the "why" behind the numbers. It directly captures whether users feel the feature improved their experience, which is the core goal.


## Guardrail Metrics

These metrics are crucial for ensuring the new feature doesn't have unintended negative consequences. They act as "do no harm" checks.
1.	Feature Abandonment Rate:
- What it is: The percentage of users who start the feature's workflow but drop off before completion.
- Why it matters: This is the inverse of completion rate. A high abandonment rate is a major red flag indicating a confusing interface, a non-obvious value proposition, or a bug that's causing frustration.
2.	Overall Churn Rate:
- What it is: The percentage of users who stop using your product altogether. You would specifically look at the churn rate of the cohort of users exposed to the new feature.
- Why it matters: While a single feature shouldn't cause mass churn, a poorly implemented or confusing feature can be the final straw for a frustrated user. A spike in churn among the exposed cohort is a critical warning.
3.	Support Tickets and Bug Reports:
- What it is: The number of support requests or bug reports directly related to the new feature.
- Why it matters: This is a real-time indicator of user pain points. A high volume of tickets immediately post-launch signals a broken user experience or a lack of clear instructions.
4.	Usage of Existing Core Features:
- What it is: Track whether the introduction of the new feature leads to a decline in usage of other key features.
- Why it matters: This helps you understand if the new feature is an improvement or simply a distraction. If it cannibalizes a more valuable core workflow without providing a net gain in user value, it could be a detrimental addition.
